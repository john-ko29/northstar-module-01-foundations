```python
import numpy as np
import pandas as pd

skyline = pd.read_csv("../lesson-1-2-types-of-data/skyline_enrollments.csv")
data = skyline["hours_studied"].values
n = len(data)
B = 10000

sample_mean = skyline["hours_studied"].mean()
sample_std = skyline["hours_studied"].std()

# Standard error of the mean
standard_error = sample_std / np.sqrt(n)

# Critical value for 95% confidence
# Note: for small samples (n < 30), we'd use t-distribution.
# n=50 is a borderline case; both z and t are reasonable. We'll use z first.
z = 1.96

# Margin of error
margin_of_error = z * standard_error

# Confidence interval
ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

np.random.seed(42)

boot_means = np.array([
    np.random.choice(data, size=n, replace=True).mean()
    for _ in range(B)
])

ci_mean = np.percentile(boot_means, [2.5, 97.5])

print(f"Sample size: {n}")
print(f"Sample mean: {sample_mean:.2f} hours")
print()
print(f"Classical 95% CI for the mean:   ({ci_lower:.2f} hours, {ci_upper:.2f} hours)")
print(f"Bootstrap 95% CI for the mean:   ({ci_mean[0]:.2f} hours, {ci_mean[1]:.2f} hours)")
```

## Results

Yes both CI are similiar with difference between .03 to .1

```python
boot_medians = np.array([
    np.median(np.random.choice(data, size=n, replace=True))
    for _ in range(B)
])

ci_median = np.percentile(boot_medians, [2.5, 97.5])

print(f"Sample median: {np.median(data):.2f} hours")
print(f"Bootstrap 95% CI for the median:   ({ci_median[0]:.2f} hours, {ci_median[1]:.2f} hours)")
```

## Results

The sample median falls between the bootstrap 95% CI for the median. 

```python
# Bootstrap 95th percentile
boot_p95 = np.array([
    np.percentile(np.random.choice(data, size=n, replace=True), 95)
    for _ in range(B)
])
sample_p95 = np.percentile(data, 95)
lo, hi = np.percentile(boot_p95, [2.5, 97.5])   # percentile-method CI

print(f"Sample 95th percentile: {sample_p95:.2f} hours")
print(f"Bootstrap 95% CI: ({lo:.2f}, {hi:.2f})")
```

## Result

Median CI tend to sit in the center of the dataset, resulting in more data to work on. While 95% percentile sits at the tail of the data where data is more sparse. 

```python
import plotly.express as px

bootstrap_df = pd.DataFrame({
    "boot_mean": boot_means,
    "boot_median": boot_medians,
})

fig_mean = px.histogram(
    bootstrap_df,
    x="boot_mean",
    nbins=50,
    title="Bootstrap Distribution: Mean Total Spent",
)
fig_mean.add_vline(x=ci_mean[0], line_dash="dash", annotation_text="2.5%")
fig_mean.add_vline(x=ci_mean[1], line_dash="dash", annotation_text="97.5%")
fig_mean.show()

fig_median = px.histogram(
    bootstrap_df,
    x="boot_median",
    nbins=50,
    title="Bootstrap Distribution: Median Total Spent",
)
fig_median.add_vline(x=ci_median[0], line_dash="dash", annotation_text="2.5%")
fig_median.add_vline(x=ci_median[1], line_dash="dash", annotation_text="97.5%")
fig_median.show()
```
