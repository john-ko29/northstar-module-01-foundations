```python
import numpy as np
import pandas as pd
from scipy import stats
import plotly.express as px
```

```python
skyline = pd.read_csv("../lesson-1-2-types-of-data/skyline_enrollments.csv")

grade_map = {"F": 0, "D": 1, "C": 2, "B": 3, "A": 4}

skyline["final_grade"] = skyline["final_grade"].astype(str).str.strip().str.upper()

skyline["final_grade_num"] = skyline["final_grade"].map(grade_map)

# Pearson correlation between total spent and visits
pearson_r, pearson_p = stats.pearsonr(
    skyline["hours_studied"],
    skyline["final_grade_num"],
)

# Spearman correlation between the same two columns
spearman_r, spearman_p = stats.spearmanr(
    skyline["hours_studied"],
    skyline["final_grade_num"],
)

print(f"Pearson r:  {pearson_r:.3f} (p = {pearson_p:.4f})")
print(f"Spearman r: {spearman_r:.3f} (p = {spearman_p:.4f})")
```

```python
fig = px.scatter(
    skyline,
    x="hours_studied",
    y="final_grade_num",
    title="Hours Studied vs. Final Grade",
    trendline="ols",
)
fig.show()
```

## Explanation

Based on the data given, the correlation do not agree. The relationship is not plausible casual since the r is less than 0.1. We need to increase r to claim casuation. 

## Confounders

1. Course difficulty 

The course may not very hard and the grading may be very lenient. Studies may easily get high grade despite the hours studied if the class grading is very easy.

2. Prior Skills

There could be incident where may experienced students would enroll these courses. They might excel with high grades despites hours put in compared to other students with no prior experience




## A/B test design.

1. Treatment and control conditions

Treatment - The students in this group will be have the built-in study planner.

Control - Group will not have built-in study planner and only the functions that already exist with the school

2. The randomization unit (per-student, per-enrollment, per-cohort?)

It will per-student. This will let students from all different courses and enrollment years. This will allow the students in both groups to have similiar years and courses.

3. Outcome metric

How many students have higher final grades. How many students from the treatment group compared to the control group have higher amount of Bs or As.

4. Minimum Effect Size

5%. Since 5% would be 1/20 of students will be  end with B or better. Since this is a planner, it would be too drastic of an improvement
