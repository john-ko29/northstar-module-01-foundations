```python
import numpy as np
import pandas as pd

bakery = pd.read_csv("../lesson-1-2-types-of-data/bakery_customers.csv")
data = bakery["total_spent_usd"].values
n = len(data)
B = 10000

np.random.seed(42)

boot_means = np.array([
    np.random.choice(data, size=n, replace=True).mean()
    for _ in range(B)
])

boot_medians = np.array([
    np.median(np.random.choice(data, size=n, replace=True))
    for _ in range(B)
])

ci_mean = np.percentile(boot_means, [2.5, 97.5])
ci_median = np.percentile(boot_medians, [2.5, 97.5])

print(f"Sample size: {n}")
print(f"Sample mean: ${data.mean():.2f}")
print(f"Sample median: ${np.median(data):.2f}")
print()
print(f"Bootstrap 95% CI for the mean:   (${ci_mean[0]:.2f}, ${ci_mean[1]:.2f})")
print(f"Bootstrap 95% CI for the median: (${ci_median[0]:.2f}, ${ci_median[1]:.2f})")
```

```python
import plotly.express as px

bootstrap_df = pd.DataFrame({
    "boot_mean": boot_means,
    "boot_median": boot_medians,
})

fig_mean = px.histogram(
    bootstrap_df,
    x="boot_mean",
    nbins=50,
    title="Bootstrap Distribution: Mean Total Spent",
)
fig_mean.add_vline(x=ci_mean[0], line_dash="dash", annotation_text="2.5%")
fig_mean.add_vline(x=ci_mean[1], line_dash="dash", annotation_text="97.5%")
fig_mean.show()

fig_median = px.histogram(
    bootstrap_df,
    x="boot_median",
    nbins=50,
    title="Bootstrap Distribution: Median Total Spent",
)
fig_median.add_vline(x=ci_median[0], line_dash="dash", annotation_text="2.5%")
fig_median.add_vline(x=ci_median[1], line_dash="dash", annotation_text="97.5%")
fig_median.show()
```
