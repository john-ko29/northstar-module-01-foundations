```python
import pandas as pd
import numpy as np
```

```python
skyline = pd.read_csv("../lesson-1-2-types-of-data/skyline_enrollments.csv")
skyline.head(10)
```

```python
course_name_mode = skyline["course_name"].mode()
print(f"Most common course: {course_name_mode.values[0]}")
```

```python
completion_status_mode = skyline["completion_status"].mode()
print(f"Most common completion status: {completion_status_mode.values[0]}")
```

```python
enrollment_year_mode = skyline["enrollment_year"].mode()
print(f"Most common enrollment year: {enrollment_year_mode.values[0]}")
```

```python
enrollment_year_median = skyline["enrollment_year"].median()
print(f"Median enrollment year: {enrollment_year_median}")
```

```python
enrollment_year_mean = skyline["enrollment_year"].mean()
print(f"Mean enrollment year: {enrollment_year_mean}")
```

```python
hours_studied_mean = skyline["hours_studied"].mean()
print(f"Mean hours studied: {hours_studied_mean:.2f}")
```

```python
hours_studied_median = skyline["hours_studied"].median()
print(f"Median hours studied: {hours_studied_median:.2f}")
```

The closer the median and mean are to each other, the more symmetric and less skewed the distribution is. No extreme outliers either. 

```python
hours_studied_mode = skyline["hours_studied"].mode()
print(f"Most common hours studied: {hours_studied_mode.values}")
```

```python
studied_min = skyline["hours_studied"].min()
studied_max = skyline["hours_studied"].max()
studied_range = studied_max - studied_min
studied_variance = skyline["hours_studied"].var()
studied_std = skyline["hours_studied"].std()

print(f"Min: {studied_min:.2f}")
print(f"Max: {studied_max:.2f}")
print(f"Range: {studied_range:.2f}")
print(f"Variance: {studied_variance:.2f} (squared hours)")
print(f"Standard deviation: {studied_std:.2f}")
```

```python
spent_q1 = skyline["hours_studied"].quantile(0.25)
spent_q3 = skyline["hours_studied"].quantile(0.75)
spent_iqr = spent_q3 - spent_q1

print(f"Q1 (25th percentile): {spent_q1:.2f}")
print(f"Q3 (75th percentile): {spent_q3:.2f}")
print(f"IQR: {spent_iqr:.2f}")
```

```python
spent_p90 = skyline["hours_studied"].quantile(0.90)
spent_p95 = skyline["hours_studied"].quantile(0.95)
spent_p99 = skyline["hours_studied"].quantile(0.99)

print(f"90th percentile: ${spent_p90:.2f}")
print(f"95th percentile: ${spent_p95:.2f}")
print(f"99th percentile: ${spent_p99:.2f}")
```

```python
final_grade_mode = skyline["final_grade"].mode()
print(f"Mode final grade: {final_grade_mode.values[0]}")
```

Having a mean for ordinal data doesn't have guaranteed equal gaps between values. This would be considered a category error.
